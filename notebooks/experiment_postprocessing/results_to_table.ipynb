```python
import numpy as np
import pandas as pd

def mean_pm_std_to_latex(
    df_mean: pd.DataFrame,
    df_std: pd.DataFrame,
    ndigits: int = 3,
    pm: str = r"\pm",
    na_str: str = "",
    # column organization
    sort_cols_by_missing: bool = True,
    first_col: str | None = "NME",   # force this to be first metric column (set None to disable)
    last_col: str = "FlowPK",        # force this to be rightmost metric column
    # row organization
    missing_sort_col: str | None = "NME",
    missing_first: bool = False,     # <-- default changed
    # highlighting
    highlight_best: bool = True,
) -> str:
    df_std = df_std.reindex_like(df_mean)

    mean_num = df_mean.apply(pd.to_numeric, errors="coerce")
    std_num  = df_std.apply(pd.to_numeric, errors="coerce")

    cols_all = list(df_mean.columns)

    # Pull out forced first/last cols (if present)
    forced_first = [first_col] if (first_col is not None and first_col in cols_all) else []
    forced_last  = [last_col]  if (last_col in cols_all and last_col not in forced_first) else []

    remaining = [c for c in cols_all if c not in forced_first and c not in forced_last]

    if sort_cols_by_missing and remaining:
        miss_rate = mean_num[remaining].isna().mean(axis=0)
        remaining = list(miss_rate.sort_values(kind="stable").index)

    ordered_cols = forced_first + remaining + forced_last

    df_mean = df_mean[ordered_cols]
    df_std  = df_std[ordered_cols]
    mean_num = mean_num[ordered_cols]
    std_num  = std_num[ordered_cols]

    # Row order: move rows with missing in missing_sort_col to beginning/end
    if missing_sort_col is not None and missing_sort_col in df_mean.columns:
        miss = mean_num[missing_sort_col].isna()
        order = miss.sort_values(ascending=missing_first, kind="stable").index
        df_mean, df_std = df_mean.loc[order], df_std.loc[order]
        mean_num, std_num = mean_num.loc[order], std_num.loc[order]

    # Best (row-wise min) mask among numeric columns
    numeric_cols = [c for c in df_mean.columns if not mean_num[c].isna().all()]
    best_mask = None
    if highlight_best and numeric_cols:
        row_min = mean_num[numeric_cols].min(axis=1, skipna=True)
        tol = 10 ** (-(ndigits + 2))
        best_mask = mean_num[numeric_cols].sub(row_min, axis=0).abs().le(tol)

    fmt = f"{{:.{ndigits}f}}"
    df_disp = df_mean.copy()

    for c in df_mean.columns:
        if c not in numeric_cols:
            continue

        mcol = mean_num[c].to_numpy()
        scol = std_num[c].to_numpy()
        is_best = (
            best_mask[c].to_numpy()
            if (best_mask is not None and c in best_mask.columns)
            else np.zeros_like(mcol, dtype=bool)
        )

        out = []
        for m, s, b in zip(mcol, scol, is_best):
            if pd.isna(m):
                out.append(na_str)
                continue

            content = fmt.format(float(m)) if pd.isna(s) else f"{fmt.format(float(m))} {pm} {fmt.format(float(s))}"
            out.append(rf"$\mathbf{{{content}}}$" if b else rf"${content}$")

        df_disp[c] = out

    return df_disp.to_latex(escape=False, index=True, na_rep=na_str)

# ---- usage ----
df_mean = df_merged.set_index("drug")
df_std  = df_merged_std.set_index("drug")

latex_str = mean_pm_std_to_latex(
    df_mean, df_std,
    first_col="NME",      # NME becomes leftmost metric column
    last_col="FlowPK",    # FlowPK becomes rightmost metric column
    missing_sort_col="NME",
    missing_first=True,  # missing rows pushed to the end
)
print(latex_str)
```
